## Weights

Lightship weight

```python
W_hull = structural_weight(C_b, LBP, B, D, d)
print(f"Hull weight: {W_hull:.2f} tons")
```

```python
W_maq = machinery_weight(pot_max_cont)
print(f"Machinery weight: {W_maq:.2f} tons")
```

```python
W_outfit = outfit_weight(LBP, B)
print(f"Outfit weight: {W_outfit:.2f} tons")
```

```python
W_LS = lightship_weight(W_hull, W_maq, W_outfit)
print(f"Lightship weight: {W_LS:.2f} tons")
```

DWT_total

```python
DWT_fuel = fuel_weight(pot_max_cont, autonomia, Vs)
print(autonomia, "km")
print(f"Fuel weight: {DWT_fuel:.2f} tons")
```

```python
DWT_water = fresh_water_and_provisions_weight(n_crew, duracao_viagem)
print(f"{duracao_viagem:.1f} dias")
print(f"Fresh water and provisions weight: {DWT_water:.2f} tons")
```

```python
DWT_crew = crew_weight(n_crew)
print(f"Crew weight: {DWT_crew:.2f} tons")
```

```python
DWT_total = total_deadweight(DWT, DWT_fuel, DWT_water, DWT_crew)
print(f"Total deadweight: {DWT_total:.2f} tons")
```

```python
print(f"Total ship weight: {DWT_total + W_LS:.2f} tons")
```

Centers

```python
LCG_hull = longitudinal_center_of_gravity_hull(LCB)
VCG_hull = vertical_center_of_gravity_hull(LBP, D, C_b)
print("Hull")
print(
    f"Longitudinal center of gravity: {LCG_hull:.2f} {'AV' if LCG_hull > 0 else 'AR'} m"
)
print(f"Vertical center of gravity: {VCG_hull:.2f} m")
```

```python
LCG_machinery = longitudinal_center_of_gravity_machinery(LBP, B, d)
VCG_machinery = vertical_center_of_gravity_machinery(LBP, B, d, D)
print("Machinery")
print(f"Longitudinal center of gravity: {LCG_machinery:.2f} m")
print(f"Vertical center of gravity: {VCG_machinery:.2f} m")
```

```python
LCG_outfit = longitudinal_center_of_gravity_outfit_watson(LCG_machinery, 25, LBP)
VCG_outfit = vertical_center_of_gravity_outfit_kupras(LBP, D)
print("Outfit")
print(f"Longitudinal center of gravity: {LCG_outfit:.2f} m")
print(f"Vertical center of gravity: {VCG_outfit:.2f} m")
```

Integration

```python
LCG_ship = (W_hull * LCG_hull + W_maq * LCG_machinery + W_outfit * LCG_outfit) / (
    W_hull + DWT_total
)

VCG_ship = (W_hull * VCG_hull + W_maq * VCG_machinery + W_outfit * VCG_outfit) / (
    W_hull + DWT_total
)

print(f"LCG_ship = {LCG_ship:.2f} m")
print(f"VCG_ship = {VCG_ship:.2f} m")
```
