## **Exemplo Prático com Pomegranate:**
```python
from pomegranate import MarkovChain
import numpy as np

# Definir matriz de transição
transicao = np.array([
    [0.7, 0.2, 0.1],  # Estado A
    [0.3, 0.5, 0.2],  # Estado B
    [0.2, 0.3, 0.5]   # Estado C
])

# Criar cadeia de Markov
modelo = MarkovChain.from_matrix(transicao)

# Simular trajetórias
trajetoria = modelo.sample(50)
print(trajetoria)
```
