Suavizamento exponencial

```python
serie = [23.56, 26.07, 21.86, 31.24, 23.65, 23.88, 26.41, 22.67]

def suav_exp(serie, alpha):
    res = [serie[0]]
    for n in range(1, len(serie)):
        res.append(alpha * serie[n] + (1 - alpha) * res[n - 1])
    return res
```

```python
resultado = []
alp = 0.1
resultado = suav_exp(serie,alp)
print(resultado[-1])
```
