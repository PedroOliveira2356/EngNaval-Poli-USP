```python
import cmath    # Funções Matemáticas com Números Complexos
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('ggplot')
```

```python
z = 3 + 4j  # Observe o 'j' para a parte imaginária
print(z)    # Saída: (3+4j)

w = complex(3, 4)  # Primeiro argumento: parte real, segundo: parte imaginária
print(w)           # Saída: (3+4j)

y = complex('3+4j')
print(y)  # Saída: (3+4j)
```

```python
z = 3 + 4j

print(z.real)  # Parte real: 3.0
print(z.imag)  # Parte imaginária: 4.0
```

```python
z1 = 2 + 3j
z2 = 4 + 5j
print(z1 + z2)  # Saída: (6+8j)
```

```python
print(z1 * z2)  # Saída: (-7+22j)
```

```python
print(z1 / z2)  # Saída: (0.5609756097560976+0.0487804878048781j)
```

```python
print(z1 ** 2)  # Saída: (-5+12j)
```

```python
z = 1 + 1j

# Raiz quadrada
print(cmath.sqrt(z))  # Saída: (1.09868411346781+0.45508986056222733j)

# Funções trigonométricas
print(cmath.sin(z))   # Saída: (1.2984575814159773+0.6349639147847361j)
print(cmath.cos(z))   # Saída: (0.8337300251311491-0.9888977057628651j)
print(cmath.tan(z))   # Saída: (0.2717525853195117+1.0839233273386946j)

# Exponencial
print(cmath.exp(z))   # Saída: (1.4686939399158851+2.2873552871788423j)

# Logaritmo
print(cmath.log(z))   # Saída: (0.34657359027997264+0.7853981633974483j)

# Fase (ângulo)
print(cmath.phase(z)) # Saída: 0.7853981633974483 (radianos, π/4)

# Coordenadas polares
print(cmath.polar(z)) # Saída: (1.4142135623730951, 0.7853981633974483) (módulo, fase)
```

```python
# De retangular para polar
z = 1 + 1j
modulo, fase = cmath.polar(z)
print(f"Módulo: {modulo}, Fase: {fase} rad")

# De polar para retangular
z_novo = cmath.rect(modulo, fase)
print(z_novo)  # Saída: (1.0000000000000002+1.0000000000000002j)
```

```python
z = 3 + 4j
conjugado = z.conjugate()
print(conjugado)  # Saída: (3-4j)
```

```python
z = 3 + 4j
print(type(z))  # Saída: <class 'complex'>
```

```python
def solve_quadratic(a, b, c):
    discriminante = (b**2) - (4*a*c)
    raiz1 = (-b + cmath.sqrt(discriminante)) / (2*a)
    raiz2 = (-b - cmath.sqrt(discriminante)) / (2*a)
    return raiz1, raiz2

# Exemplo: x² + 2x + 5 = 0
print(solve_quadratic(1, 2, 5))  # Saída: ((-1+2j), (-1-2j))
```

```python
# Exemplo simples de FFT (Fast Fourier Transform)
sinal = np.array([1, 2, 3, 4])
fft_resultado = np.fft.fft(sinal)
print(fft_resultado)
# Saída: [10.+0.j -2.+2.j -2.+0.j -2.-2.j]
```

```python
# Criar um array de números complexos
theta = np.linspace(0, 2*np.pi, 100)
z = np.exp(1j * theta)  # Círculo unitário

# Plotar
plt.figure(figsize=(6, 6))
plt.plot(z.real, z.imag)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.title('Círculo Unitário no Plano Complexo')
plt.xlabel('Parte Real')
plt.ylabel('Parte Imaginária')
plt.axis('equal')
plt.show()
```
