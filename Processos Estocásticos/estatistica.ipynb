```python
import numpy as np
from scipy import stats

# Gera uma amostra aleatória a partir de uma distribuição normal
np.random.seed(42)
```

```python
observado = [3, 12, 5, 5]
esperado = np.ones(len(observado)) * np.mean(observado)

m = 2

# Teste de aderência Qui-Quadrado
stats.chisquare(f_obs=observado, f_exp=esperado, ddof=m)
```

```python
p_val = stats.chi2.sf(7.48, 1)
print(f"p-value: {p_val}")
```

```python
# média=0, desvio padrão=1, tamanho=100
sample = np.random.normal(loc=0, scale=1, size=100)

# Realiza o teste de Kolmogorov-Smirnov
ks_statistic, p_value = stats.kstest(sample, "norm")

if p_value > 0.05:
    print("Não rejeitamos H0: a amostra segue distribuição normal.")
else:
    print("Rejeitamos H0: a amostra não segue distribuição normal.")
```
