```python
from math import pi

import sympy as sp


def resolver_equacao(equacao, valores=None):
    """
    equacao: sp.Eq, expressão sympy (assumida == 0), ou string (ex: "a*x + b - c")
    valores: dict {simbolo_ou_nome: valor} a substituir na equação

    Substitui os valores conhecidos e resolve para a única variável livre
    restante, seja ela qual for.
    """
    if isinstance(equacao, str):
        equacao = sp.sympify(equacao)
    if not isinstance(equacao, sp.Eq):
        equacao = sp.Eq(equacao, 0)

    valores = valores or {}
    valores = {(sp.Symbol(k) if isinstance(k, str) else k): v
               for k, v in valores.items()}

    eq_substituida = equacao.subs(valores)
    livres = eq_substituida.free_symbols

    if len(livres) != 1:
        raise ValueError(
            f"Esperava exatamente 1 variável livre após a substituição, "
            f"encontrou {len(livres)}: {livres}"
        )

    variavel = livres.pop()
    solucoes = sp.solve(eq_substituida, variavel)
    return variavel, solucoes


def area_pistao(D):
    return pi * (D/2)**2


def pot_motor(pme_f, A_p, L, z, N, x):
    """
    Calcula a potência do motor.

    pme_f: pressão média efetiva do motor (Pa)
    A_p: área do pistão (m²)
    L: curso do pistão (m)
    z: número de cilindros
    N: rotação do motor (rps)
    x: fator de conversão (1 para motores de 2 tempos
                           2 para motores de 4 tempos)

    Retorna a potência do motor (kW).
    """
    return (pme_f * A_p * L * z * N / x) / 1000
```

```python
# Encontrar pme
pot_m, pme_f, A_p, L, z, N, x = sp.symbols('pot_m pme_f A_p L z N x')
eq = sp.Eq(pot_m, pme_f * A_p * L * z * N / x)

var, sol = resolver_equacao(eq, {pot_m: 39500*0.85, A_p: area_pistao(0.85), L: 3.145, z: 1, N: 76/60, x: 1})
print(var, '=', sol[0])
```
