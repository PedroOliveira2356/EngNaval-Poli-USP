```python
import matplotlib
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
matplotlib.style.use('ggplot')
```

```python
# Constantes
g = 9.81                    # gravidade; m/s^2
L = 1.5                     # comprimento da balsa; m
r = 0.15                    # raio da boia de isopor; m
b = 1.                      # largura da balsa; m
gamma_iso = 15 * g          # peso especifico do isopor; N/m^3
gamma_agua = 1000 * g       # peso especifico da agua; N/m^3
d = 0.375                   # distancia 'd'; m
theta_8rad = 8*np.pi/180    # theta barra; rad

Wmin = 600                                      # peso minimo da carga; N
Wmax = 2*np.pi*r**2*b*(gamma_agua - gamma_iso)  # peso maximo da carga; N
t0 = 0; t8 = np.tan(theta_8rad)                 # tan theta; -
BM = (b*L**3*gamma_agua)/12                     # raio metacentrico; m
```

```python
peso_iso = 2*np.pi*r**2*b * gamma_iso   # peso das 2 boias de isopor; N
peso_iso
```

```python
V_iso = 2*np.pi*r**2*b          # volume das 2 boias de isopor; m^3
V_iso
```

```python
v0 = g*(BM-0.0429*peso_iso)     # valores auxiliares
vw = (-0.9-0.0429)*g

def ttheta():                   # calculo de tan theta para valores de W
    W = float(input('Valor para W: '))
    return (W*d)/(v0+vw*W)

ttheta()
```

```python
# Rotina para gerar graficos
W = np.linspace(600, Wmax*2)
d = L/2
tan_theta = (W*d)/(27057.6483-9.249849*W)

fig, ax1 = plt.subplots(1, 1, sharex=True, figsize=(6, 4))
ax1.fill_between([Wmin, Wmax], [t8, t8], alpha = 0.5, )
plt.plot([530, 2700], [t8, t8], '--c')

plt.xlabel('Peso da carga (W)')
plt.ylabel('tan 'r'$\theta$')

plt.plot(W, tan_theta, 'b');
```
